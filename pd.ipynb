```
import pandas as pd
#  to use it in the file
```

```
#core Data structures in pandas are Series and Data Frames
# Series is just like a column with index
# to define a Series in pandas use the function Series ()
# as input it takes a list []
s = pd.Series([10,20,30,40])
print(s)
```

```
# Here the indexes are not by default 0,1,2,3 but instead are a,b,c,d
s=pd.Series([10,20,30,40],index=["a","b","c","d"])
print(s)
```

```
# Data frame in pandas is just a 2D Data structure
# use the function DataFrame()
# it takes as an argument a dictionary {}
# each key is the name of the column and the values of the column is a Series
data = {
    "Names":["Ali","Omar","John Doe"],
    "Ages":[29,28,20],
    "Majors":["CS","ENG","MED"]
}
# create a Data frame
df= pd.DataFrame(data)
print(df)
```

###### Filtering Data

```
# Print only the rows where the Age > 22
# The condition is df ["Age"] > 22  -> this condition will come inside df [(df ["Age"] > 22 )]
print(df[df["Age"] >22])
```

```
# print only the rows where Age > 22 & City == Berlin
# 1st condition df["Age“] > 22
# 2nd condition df ["City"] == "Berlin"
# put them together using & -> (df["Age“] > 22 ) & (df ["City"] == "Berlin" )
# now the condition will come inside df []
print(df [   (df["Age"] > 22) & (df["City"]=="Berlin") ])
```

```
#print thr rows where city  Berlin or Hamburg
print (df[ ( df ["City"] =="Berlin") | (df["City"] == "Hamburg")])
```

```
# isin method  -> print the rows where the city = Hamburg | Berlin | NYC , you can also us multiple condition , but this is faster
print(df[df["City"]. isin(["Hamburg","NYC","Berlin"])])
```

###### cleaning Data

```
df2=pd.DataFrame({
    "Name":["Ali","Sara",None,None,"Ali"],
    "Age":[22,None,23,21,22],
    "Grade":[1.3,2.0,1.7,None,1.3]
})
print(df2)
```

```
# To find the None values , that does not exist , use isnull, it will print True of False , True if does not exist
print(df2.isnull())
print(df2.isnull().sum()) # How many value does not exist in each column for example in Name -> 2 , Age -> 1 , Grade -> 1
```

```
# clear the frame by dropping the None Values
df_clean= df2.dropna() # it will drop all the rows where None exists
print(df_clean)
# drop all None value but only for the column Age
print(".........")
df_clean = df2.dropna(subset=["Age"]) # only the row with index =1 will be dropped
print(df_clean)
```
